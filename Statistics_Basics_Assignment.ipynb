# Question 1: What is the Difference Between Descriptive Statistics and Inferential Statistics? Explain with Examples.

## Introduction

**Statistics** is the branch of mathematics that deals with collecting, organizing, analyzing, interpreting, and presenting data.

Statistics is mainly divided into two types:

1. **Descriptive Statistics**
2. **Inferential Statistics**

The main difference is that **descriptive statistics summarizes the data that has been collected**, while **inferential statistics uses sample data to make conclusions or predictions about a larger population**.

---

# 1. Descriptive Statistics

**Descriptive statistics** is used to summarize and describe the main characteristics of a dataset.

It does not make predictions or generalizations beyond the data being analyzed.

### Common Methods

* Mean
* Median
* Mode
* Range
* Variance
* Standard deviation
* Minimum and maximum
* Frequency tables
* Charts and graphs

### Example

Suppose the marks of five students are:

```text
60, 70, 75, 80, 90
```

The mean is:

$$
Mean = \frac{60+70+75+80+90}{5}=75
$$

Therefore, we can say that the **average mark of these five students is 75**.

This is descriptive statistics because we are only summarizing the given five students.

### Real-World Example

A company analyzes the monthly sales of its 100 employees and finds:

* Average sales = ₹50,000
* Highest sales = ₹90,000
* Lowest sales = ₹20,000
* Standard deviation = ₹12,000

These values describe the existing sales data.

---

# 2. Inferential Statistics

**Inferential statistics** is used to draw conclusions, make predictions, or make decisions about a **population based on sample data**.

Since it is usually difficult or expensive to collect data from an entire population, researchers study a sample and use statistical methods to infer information about the population.

### Common Methods

* Estimation
* Confidence intervals
* Hypothesis testing
* Regression analysis
* Correlation
* t-test
* Chi-square test
* ANOVA

### Example

Suppose a university has **10,000 students**.

Instead of asking every student about their average study time, researchers select a random sample of **500 students**.

If the sample shows that students study an average of **3.5 hours per day**, researchers may use inferential statistics to estimate the average study time of all 10,000 students.

Here:

```text
Population → 10,000 students
Sample → 500 students
Sample result → 3.5 hours
Inference → Estimate for the entire population
```

This is inferential statistics because the sample is being used to make a conclusion about the larger population.

---

# Difference Between Descriptive and Inferential Statistics

| Basis          | Descriptive Statistics                       | Inferential Statistics                               |
| -------------- | -------------------------------------------- | ---------------------------------------------------- |
| Purpose        | Summarizes and describes data                | Draws conclusions about a population                 |
| Data Used      | Usually the collected dataset                | Usually a sample from a population                   |
| Prediction     | Does not primarily predict                   | Can make predictions/estimates                       |
| Generalization | Does not generalize beyond the analyzed data | Generalizes from sample to population                |
| Methods        | Mean, median, mode, range, SD                | Hypothesis testing, confidence intervals, regression |
| Example        | Average marks of a class                     | Estimating average marks of an entire university     |
| Main Question  | "What does the data show?"                   | "What can we conclude about the population?"         |

---

# Simple Example to Understand the Difference

Suppose a company surveys **100 customers**.

### Descriptive Statistics

The company finds:

$$
Average\ satisfaction = 4.2/5
$$

This describes the satisfaction of the **100 surveyed customers**.

### Inferential Statistics

The company uses those 100 customers as a sample to estimate the satisfaction of **all 50,000 customers**.

This is inferential statistics.

---

# Relationship Between the Two

Descriptive and inferential statistics are often used together:

```text
Collect Data
     ↓
Descriptive Statistics
     ↓
Summarize the Sample
     ↓
Inferential Statistics
     ↓
Make Conclusions About Population
```

For example, a researcher may first calculate the **mean and standard deviation** of a sample and then use **hypothesis testing or confidence intervals** to draw conclusions about the population.

---

# Key Point

The easiest way to remember the difference is:

> **Descriptive statistics describes what the data contains, whereas inferential statistics uses data from a sample to make conclusions about a larger population.**

## Conclusion

**Descriptive statistics** focuses on organizing, summarizing, and presenting the data that has been collected. It includes measures such as mean, median, mode, range, and standard deviation.

**Inferential statistics** goes one step further by using sample data to estimate, predict, or make conclusions about a larger population. It includes techniques such as hypothesis testing, confidence intervals, and regression.

Therefore:

$$
\boxed{\text{Descriptive = Describe the data}}
$$

$$
\boxed{\text{Inferential = Infer conclusions about the population}}
$$


# Question 2: What is Sampling in Statistics? Explain the Differences Between Random and Stratified Sampling.

## Introduction

In statistics, it is often difficult, expensive, or time-consuming to collect information from every member of a population. **Sampling** provides a practical way to study a smaller group and use the results to understand the larger population.

---

# What is Sampling?

**Sampling** is the process of selecting a smaller group of individuals, called a **sample**, from a larger group, called the **population**, for the purpose of collecting and analyzing data.

### Example

Suppose a college has **5,000 students** and we want to know their average satisfaction with college facilities.

Instead of asking all 5,000 students, we may select **500 students** and collect their responses.

```text
Population
5,000 Students
      ↓
   Sampling
      ↓
Sample
500 Students
      ↓
Collect & Analyze Data
      ↓
Draw Conclusions
```

The **population** is the complete group, while the **sample** is the smaller group selected from it.

---

# What is Random Sampling?

**Random sampling**, also called **simple random sampling**, is a method in which every member of the population has an **equal or known chance of being selected**.

The selection is made randomly, so the researcher does not intentionally choose specific individuals.

### Example

A company has **1,000 employees** and wants to survey 100 employees.

The company assigns each employee an ID from 1 to 1,000 and randomly selects 100 IDs.

```text
1,000 Employees
       ↓
Random Selection
       ↓
100 Employees
```

Every employee has an equal chance of being selected.

### Advantages

* Simple to understand and implement.
* Reduces researcher selection bias.
* Provides a representative sample when properly conducted.
* Suitable when the population is relatively homogeneous.

### Limitations

* Requires a complete list of the population.
* Can be difficult for very large or geographically distributed populations.
* Important subgroups may be underrepresented by chance.

---

# What is Stratified Sampling?

**Stratified sampling** is a method in which the population is divided into smaller, relatively homogeneous groups called **strata** based on a relevant characteristic.

A sample is then selected from **each stratum**.

Strata may be based on:

* Age
* Gender
* Income group
* Department
* Education level
* Geographic region

### Example

Suppose a college has 5,000 students:

```text
Population
    ↓
 ┌──────────────┬──────────────┬──────────────┐
 ↓              ↓              ↓
1st Year       2nd Year       3rd Year
    ↓              ↓              ↓
 Sample          Sample         Sample
    └──────────────┬──────────────┘
                   ↓
             Final Sample
```

If the researcher wants every year to be represented, students can be divided into strata based on year of study and then randomly selected from each group.

For example:

| Stratum   | Population |  Sample |
| --------- | ---------: | ------: |
| 1st Year  |      2,000 |     200 |
| 2nd Year  |      1,500 |     150 |
| 3rd Year  |      1,000 |     100 |
| 4th Year  |        500 |      50 |
| **Total** |  **5,000** | **500** |

This is an example of **proportionate stratified sampling**, where the sample size from each group reflects its population size.

---

# Difference Between Random and Stratified Sampling

| Basis               | Random Sampling                                         | Stratified Sampling                                  |
| ------------------- | ------------------------------------------------------- | ---------------------------------------------------- |
| Meaning             | Selects individuals randomly from the entire population | Divides population into strata and samples from each |
| Population Division | No required division                                    | Population is divided into groups                    |
| Selection           | Random selection from the whole population              | Random selection within each stratum                 |
| Main Goal           | Give each member an equal/known chance of selection     | Ensure important subgroups are represented           |
| Representation      | Some groups may be underrepresented by chance           | Each selected stratum is represented                 |
| Complexity          | Relatively simple                                       | More complex                                         |
| Best Used When      | Population is relatively homogeneous                    | Population contains important distinct subgroups     |
| Example             | Randomly selecting 100 students from 5,000              | Selecting students from each year of study           |

---

# Simple Example

Suppose a company has:

* 600 male employees
* 400 female employees

and wants a sample of 100 employees.

### Random Sampling

The company randomly selects 100 employees from all 1,000 employees.

There is no separate requirement to select from each group.

### Stratified Sampling

The company divides employees into two strata:

```text
Stratum 1 → Male Employees
Stratum 2 → Female Employees
```

Then it selects employees from both groups, for example proportionately:

* 60 males
* 40 females

This ensures that both groups are represented.

---

# Advantages of Stratified Sampling

1. Ensures representation of important subgroups.
2. Can provide more precise estimates when strata are well chosen.
3. Useful when the population is heterogeneous.
4. Allows separate analysis of different groups.
5. Can reduce sampling variability compared with a simple random sample in suitable situations.

---

# Conclusion

**Sampling** is the process of selecting a smaller group from a larger population to collect and analyze data.

* **Random sampling** selects individuals randomly from the entire population, giving each member an equal or known chance of selection.
* **Stratified sampling** first divides the population into meaningful groups or strata and then selects a sample from each group.

The key difference is:

$$
\boxed{\text{Random Sampling → Random selection from the whole population}}
$$

$$
\boxed{\text{Stratified Sampling → Divide into groups → Randomly sample from each group}}
$$

Therefore, **random sampling** is useful for simple and unbiased selection, while **stratified sampling** is particularly useful when ensuring representation of important subgroups is important.

# Question 3: Define Mean, Median, and Mode. Explain Why These Measures of Central Tendency Are Important.

## Introduction

**Measures of central tendency** are statistical measures used to identify the **central or typical value** of a dataset. They provide a single value that represents the overall distribution of the data.

The three most common measures of central tendency are:

1. **Mean**
2. **Median**
3. **Mode**

---

# 1. Mean

The **mean** is the arithmetic average of all the values in a dataset.

### Formula

$$
Mean = \frac{\text{Sum of all observations}}{\text{Number of observations}}
$$

or

$$
\bar{x} = \frac{\sum x}{n}
$$

where:

* \(\bar{x}\) = Mean
* \(\sum x\) = Sum of all observations
* \(n\) = Number of observations

### Example

Consider the following marks:

```text
10, 20, 30, 40, 50
```

$$
Mean = \frac{10+20+30+40+50}{5}
$$

$$
Mean = \frac{150}{5}=30
$$

Therefore, the **mean is 30**.

### Importance of Mean

* Uses all observations in the dataset.
* Easy to calculate and understand.
* Useful for comparing different datasets.
* Widely used in further statistical calculations.
* Suitable for numerical data that does not contain extreme outliers.

### Limitation

The mean can be strongly affected by **extreme values (outliers)**.

For example:

```text
10, 20, 30, 40, 500
```

The value 500 greatly increases the mean, even though most observations are much smaller.

---

# 2. Median

The **median** is the middle value of a dataset after arranging the observations in **ascending or descending order**.

### Case 1: Odd Number of Observations

For:

```text
10, 20, 30, 40, 50
```

The middle value is:

$$
Median = 30
$$

### Case 2: Even Number of Observations

For:

```text
10, 20, 30, 40
```

There are two middle values: 20 and 30.

Therefore:

$$
Median = \frac{20+30}{2}=25
$$

### Importance of Median

* Less affected by extreme values than the mean.
* Useful for skewed data.
* Suitable for income, salary, house prices, and other datasets containing outliers.
* Provides a good representation of the central position of ordered data.

### Example

Consider salaries:

```text
₹20,000, ₹25,000, ₹30,000, ₹35,000, ₹5,00,000
```

The very high salary of ₹5,00,000 affects the mean considerably.

The median remains:

$$
Median = ₹30,000
$$

Therefore, the median may better represent the typical salary in this example.

---

# 3. Mode

The **mode** is the value that occurs **most frequently** in a dataset.

### Example

Consider:

```text
2, 3, 3, 4, 5, 3, 6
```

The number **3** occurs most frequently.

Therefore:

$$
Mode = 3
$$

### Importance of Mode

* Easy to identify.
* Useful for categorical data.
* Shows the most common value or category.
* Not affected by extreme values.
* Useful in business and market analysis.

### Example

A clothing store records the shirt sizes sold:

```text
M, L, M, S, M, XL, L
```

The mode is:

$$
Mode = M
$$

This tells the store that **Medium (M)** is the most frequently purchased size.

---

# Comparison of Mean, Median, and Mode

| Measure    | Definition                      | Main Advantage              | Effect of Outliers |
| ---------- | ------------------------------- | --------------------------- | ------------------ |
| **Mean**   | Arithmetic average              | Uses all observations       | Highly affected    |
| **Median** | Middle value                    | Good for skewed data        | Less affected      |
| **Mode**   | Most frequently occurring value | Useful for categorical data | Not affected       |

---

# Why Are Measures of Central Tendency Important?

## 1. Summarize Large Datasets

A large dataset can contain thousands or millions of observations. Mean, median, or mode can summarize the data using a single representative value.

---

## 2. Easy Comparison

Central tendency allows us to compare different groups.

For example:

```text
Class A Average = 72 marks
Class B Average = 65 marks
```

We can quickly see that Class A has a higher average performance.

---

## 3. Helps in Decision-Making

Businesses, governments, researchers, and organizations use central tendency to make decisions.

For example, a company can use the **median salary** to understand typical employee compensation.

---

## 4. Identifies the Typical Value

These measures help answer questions such as:

* What is the average income?
* What is the typical house price?
* What is the most common product size?
* What is the middle exam score?

---

## 5. Useful for Further Statistical Analysis

The mean is particularly important in many statistical techniques, including:

* Variance
* Standard deviation
* Correlation
* Regression
* Hypothesis testing

---

# When Should We Use Each Measure?

| Situation                           | Best Measure |
| ----------------------------------- | ------------ |
| Normally distributed numerical data | **Mean**     |
| Data contains extreme outliers      | **Median**   |
| Highly skewed data                  | **Median**   |
| Categorical data                    | **Mode**     |
| Finding most popular category       | **Mode**     |
| Comparing average numerical values  | **Mean**     |

---

# Simple Example

Consider the dataset:

```text
5, 7, 7, 8, 9
```

### Mean

$$
Mean=\frac{5+7+7+8+9}{5}=7.2
$$

### Median

The middle value is:

$$
Median=7
$$

### Mode

The most frequent value is:

$$
Mode=7
$$

Therefore:

```text
Mean   = 7.2
Median = 7
Mode   = 7
```

---

# Conclusion

**Mean, median, and mode** are the three major measures of central tendency used to represent the center or typical value of a dataset.

* **Mean** → Arithmetic average of the data.
* **Median** → Middle value after sorting the data.
* **Mode** → Most frequently occurring value.

They are important because they **summarize data, simplify comparisons, identify typical values, support decision-making, and provide a foundation for further statistical analysis**.

The easiest way to remember them is:

$$
\boxed{\text{Mean = Average}}
$$

$$
\boxed{\text{Median = Middle}}
$$

$$
\boxed{\text{Mode = Most Frequent}}
$$


# Question 4: Explain Skewness and Kurtosis. What Does a Positive Skew Imply About the Data?

## Introduction

**Skewness** and **kurtosis** are statistical measures used to describe the **shape and distribution of data**.

* **Skewness** measures the **asymmetry** of a distribution.
* **Kurtosis** describes the **tail heaviness and peakedness** of a distribution compared with a normal distribution.

These measures help us understand whether data is symmetric, tilted toward one side, or contains unusually heavy or light tails.

---

# 1. Skewness

**Skewness** measures the degree to which a probability distribution is **asymmetric** around its mean.

A perfectly symmetric distribution has approximately zero skewness.

There are three main types:

1. **Zero Skewness**
2. **Positive Skewness**
3. **Negative Skewness**

---

## A. Zero Skewness

When a distribution is approximately symmetric:

$$
Skewness \approx 0
$$

Typically:

$$
Mean \approx Median \approx Mode
$$

### Example

```text
2, 3, 4, 5, 6
```

The data is approximately symmetric around the center.

---

## B. Positive Skewness

A distribution is **positively skewed** when it has a **longer tail toward the right**.

```text
Frequency
   |
   |        ███
   |      █████
   |    ███████
   |  ███████
   | █████
   |____████________________→
                     Long right tail
```

In a positively skewed distribution, extreme high values pull the mean toward the right.

Usually:

$$
Mean > Median > Mode
$$

### Example

Income data is often positively skewed because most people may have moderate incomes while a small number of people have very high incomes.

---

## C. Negative Skewness

A distribution is **negatively skewed** when it has a **longer tail toward the left**.

Usually:

$$
Mean < Median < Mode
$$

### Example

If most students score very high on an easy exam but a few students score very low, the distribution may be negatively skewed.

---

# What Does Positive Skew Imply?

A **positive skew** implies that the distribution has a **long right tail**, meaning there are relatively more extreme or unusually high observations compared with a symmetric distribution.

It generally indicates:

* A few unusually high values are present.
* The mean is pulled upward.
* The **mean is usually greater than the median**.
* The distribution is not symmetric.
* The data may contain high-value outliers.

### Example

Consider the following incomes:

```text
₹20,000
₹22,000
₹25,000
₹27,000
₹30,000
₹2,00,000
```

Most incomes are between ₹20,000 and ₹30,000, but ₹2,00,000 is an unusually high value.

This creates a **long right tail**, resulting in positive skewness.

---

# 2. Kurtosis

**Kurtosis** describes the **tail behavior of a distribution**, particularly how much probability is concentrated in the tails relative to a normal distribution.

It is commonly used to identify whether data has relatively **heavy or light tails**.

There are three common categories:

1. **Mesokurtic**
2. **Leptokurtic**
3. **Platykurtic**

---

## A. Mesokurtic

A **mesokurtic** distribution has kurtosis similar to a normal distribution.

For Pearson's kurtosis:

$$
Kurtosis = 3
$$

For **excess kurtosis**:

$$
Excess\ Kurtosis = 0
$$

The normal distribution is the standard example.

---

## B. Leptokurtic

A **leptokurtic** distribution has **heavier tails** than a normal distribution.

It tends to have a greater tendency to produce extreme observations.

For excess kurtosis:

$$
Excess\ Kurtosis > 0
$$

---

## C. Platykurtic

A **platykurtic** distribution has **lighter tails** than a normal distribution.

It generally has fewer extreme observations compared with a normal distribution.

For excess kurtosis:

$$
Excess\ Kurtosis < 0
$$

---

# Skewness vs Kurtosis

| Feature                          | Skewness                            | Kurtosis                                          |
| -------------------------------- | ----------------------------------- | ------------------------------------------------- |
| Measures                         | Asymmetry                           | Tail heaviness                                    |
| Main focus                       | Direction and degree of imbalance   | Extreme-value/tail behavior                       |
| Value around normal distribution | 0                                   | 0 excess kurtosis                                 |
| Positive value                   | Right-skewed                        | Heavier tails                                     |
| Negative value                   | Left-skewed                         | Lighter tails                                     |
| Main use                         | Understand distribution's asymmetry | Understand tail behavior and extreme observations |

---

# Real-World Examples

### Positive Skewness

**Income distribution**

A large number of people may have moderate incomes, while a small number have extremely high incomes.

```text
Most observations → Moderate income
Few observations  → Very high income
                         ↓
                   Right tail
```

### Kurtosis

**Financial returns**

A dataset with high kurtosis can have relatively heavy tails, meaning extreme gains or losses may occur more frequently than under a normal distribution.

---

# Important Difference

It is useful to remember:

> **Skewness tells us about the direction of asymmetry, while kurtosis tells us about the behavior of the tails.**

For example:

```text
Skewness  → Is the distribution tilted left or right?
Kurtosis  → How heavy or light are the tails?
```

---

# Summary

| Concept           | Meaning                                                    |
| ----------------- | ---------------------------------------------------------- |
| **Skewness**      | Measures asymmetry of a distribution                       |
| **Positive Skew** | Long right tail; high values pull the mean upward          |
| **Negative Skew** | Long left tail; low values pull the mean downward          |
| **Zero Skew**     | Approximately symmetric distribution                       |
| **Kurtosis**      | Describes tail heaviness relative to a normal distribution |
| **Leptokurtic**   | Heavy tails                                                |
| **Mesokurtic**    | Similar to normal                                          |
| **Platykurtic**   | Light tails                                                |

## Conclusion

**Skewness** and **kurtosis** help describe the shape of a dataset beyond measures such as mean and standard deviation.

**Skewness** measures asymmetry. A **positive skew** means the distribution has a **longer right tail**, usually because of some unusually high observations, and the mean is typically greater than the median.

**Kurtosis** focuses primarily on the **tail behavior and frequency of extreme observations** compared with a normal distribution.

Therefore:

$$
\boxed{\text{Positive Skew} \rightarrow \text{Long Right Tail}}
$$

$$
\boxed{\text{Skewness} \rightarrow \text{Asymmetry}}
$$

$$
\boxed{\text{Kurtosis} \rightarrow \text{Tail Heaviness}}
$$


# Question 5: Implement a Python Program to Compute the Mean, Median, and Mode

## Given List

```python
numbers = [12, 15, 12, 18, 19, 12, 20, 22, 19, 19, 24, 24, 24, 26, 28]
```

## Python Program

```python
import statistics

# Given list of numbers
numbers = [12, 15, 12, 18, 19, 12, 20, 22, 19, 19, 24, 24, 24, 26, 28]

# Calculate mean, median, and mode
mean = statistics.mean(numbers)
median = statistics.median(numbers)
mode = statistics.mode(numbers)

# Display the results
print("Numbers:", numbers)
print("Mean:", mean)
print("Median:", median)
print("Mode:", mode)
```

## Output

```text
Numbers: [12, 15, 12, 18, 19, 12, 20, 22, 19, 19, 24, 24, 24, 26, 28]
Mean: 20.8
Median: 20
Mode: 12
```

## Explanation

### 1. Mean

The mean is calculated as:

$$
Mean = \frac{\text{Sum of all values}}{\text{Number of values}}
$$

There are **15 numbers**, and their sum is **312**.

$$
Mean = \frac{312}{15}=20.8
$$

Therefore:

$$
\boxed{Mean=20.8}
$$

---

### 2. Median

The numbers are already arranged in ascending order:

```text
12, 12, 12, 15, 18, 19, 19, 19, 20, 22, 24, 24, 24, 26, 28
```

There are **15 values**, so the middle position is:

$$
\frac{15+1}{2}=8
$$

The 8th value is **19**.

Therefore:

$$
\boxed{Median=19}
$$

---

### 3. Mode

The mode is the value that occurs most frequently.

Frequency:

* 12 → 3 times
* 19 → 3 times
* 24 → 3 times

Therefore, the dataset actually has **three modes**:

$$
\boxed{12,\ 19,\ 24}
$$

So this dataset is **multimodal**.

### Important Correction to the Basic `statistics.mode()` Example

In recent Python versions, `statistics.mode(numbers)` returns the **first mode encountered**, which is `12`. However, statistically, since **12, 19, and 24 each occur three times**, all three are modes.

To correctly display **all modes**, use `statistics.multimode()`:

```python
import statistics

numbers = [12, 15, 12, 18, 19, 12, 20, 22, 19, 19, 24, 24, 24, 26, 28]

mean = statistics.mean(numbers)
median = statistics.median(numbers)
modes = statistics.multimode(numbers)

print("Mean:", mean)
print("Median:", median)
print("Modes:", modes)
```

### Correct Output

```text
Mean: 20.8
Median: 19
Modes: [12, 19, 24]
```

## Functions Used

| Function                 | Purpose                        |
| ------------------------ | ------------------------------ |
| `statistics.mean()`      | Calculates the arithmetic mean |
| `statistics.median()`    | Calculates the median          |
| `statistics.mode()`      | Returns a mode                 |
| `statistics.multimode()` | Returns all modes              |

## Final Answer

$$
\boxed{Mean=20.8}
$$

$$
\boxed{Median=19}
$$

$$
\boxed{Mode=12,\ 19,\ 24}
$$

Since three values occur with the highest frequency, the dataset has **three modes**.


# Question 6: Compute Covariance and Correlation Coefficient Between Two Datasets

## Given Data

```python
list_x = [10, 20, 30, 40, 50]
list_y = [15, 25, 35, 45, 60]
```

We need to calculate:

1. **Covariance**
2. **Correlation coefficient**

---

# Python Program

```python
import numpy as np

# Given datasets
list_x = [10, 20, 30, 40, 50]
list_y = [15, 25, 35, 45, 60]

# Calculate covariance
covariance = np.cov(list_x, list_y)[0, 1]

# Calculate correlation coefficient
correlation = np.corrcoef(list_x, list_y)[0, 1]

# Display results
print("Covariance:", covariance)
print("Correlation Coefficient:", correlation)
```

## Output

```text
Covariance: 275.0
Correlation Coefficient: 0.949...
```

Therefore:

$$
\boxed{Covariance = 275}
$$

$$
\boxed{Correlation \approx 0.949}
$$

---

# 1. Covariance

**Covariance** measures the direction in which two variables change together.

The sample covariance formula is:

$$
Cov(X,Y)=\frac{\sum (X_i-\bar X)(Y_i-\bar Y)}{n-1}
$$

### Step 1: Calculate the Means

For \(X\):

$$
\bar X = \frac{10+20+30+40+50}{5}=30
$$

For \(Y\):

$$
\bar Y = \frac{15+25+35+45+60}{5}=36
$$

### Step 2: Calculate Deviations

|         X |  Y | X − 30 | Y − 36 |  Product |
| --------: | -: | -----: | -----: | -------: |
|        10 | 15 |    -20 |    -21 |      420 |
|        20 | 25 |    -10 |    -11 |      110 |
|        30 | 35 |      0 |     -1 |        0 |
|        40 | 45 |     10 |      9 |       90 |
|        50 | 60 |     20 |     24 |      480 |
| **Total** |    |        |        | **1100** |

Therefore:

$$
Cov(X,Y)=\frac{1100}{5-1}
$$

$$
\boxed{Cov(X,Y)=275}
$$

A positive covariance indicates that **X and Y tend to increase together**.

---

# 2. Correlation Coefficient

The **correlation coefficient** measures the strength and direction of the linear relationship between two variables.

The Pearson correlation coefficient is:

$$
r=\frac{Cov(X,Y)}{\sigma_X\sigma_Y}
$$

Its value lies between:

$$
-1 \leq r \leq 1
$$

For these datasets:

$$
\boxed{r \approx 0.949}
$$

---

# Interpretation

The correlation coefficient is approximately **0.949**, which is close to **+1**.

This indicates a **strong positive linear relationship** between `list_x` and `list_y`.

In simple terms:

> As the values of X increase, the values of Y also tend to increase.

For example:

# Question 7: Draw a Boxplot and Identify Outliers

## Given Data

```python
data = [12, 14, 14, 15, 18, 19, 19, 21, 22, 22, 23, 23, 24, 26, 29, 35]
```

A **boxplot** is a statistical graph used to show the distribution of numerical data and identify potential **outliers**.

---

# Python Program

```python
import matplotlib.pyplot as plt

# Given data
data = [12, 14, 14, 15, 18, 19, 19, 21,
        22, 22, 23, 23, 24, 26, 29, 35]

# Draw boxplot
plt.boxplot(data)

plt.title("Boxplot of Given Data")
plt.ylabel("Values")
plt.show()

# Calculate Q1 and Q3 using the median-of-halves method
data_sorted = sorted(data)

n = len(data_sorted)

# Lower and upper halves
lower_half = data_sorted[:n // 2]
upper_half = data_sorted[n // 2:]

# Calculate Q1 and Q3
q1 = (lower_half[3] + lower_half[4]) / 2
q3 = (upper_half[3] + upper_half[4]) / 2

# Calculate IQR
iqr = q3 - q1

# Calculate lower and upper fences
lower_fence = q1 - 1.5 * iqr
upper_fence = q3 + 1.5 * iqr

# Identify outliers
outliers = [x for x in data if x < lower_fence or x > upper_fence]

print("Q1:", q1)
print("Q3:", q3)
print("IQR:", iqr)
print("Lower Fence:", lower_fence)
print("Upper Fence:", upper_fence)
print("Outliers:", outliers)
```

# Question 8: Relationship Between Advertising Spend and Daily Sales

## Introduction

As a data analyst in an e-commerce company, **covariance and correlation** can be used to understand whether advertising spend and daily sales move together.

The two variables are:

```python
advertising_spend = [200, 250, 300, 400, 500]
daily_sales = [2200, 2450, 2750, 3200, 4000]
```

Here:

* **Advertising spend** = amount spent on advertising each day.
* **Daily sales** = sales generated on each day.

---

# 1. Using Covariance

**Covariance** measures the **direction of the relationship** between two variables.

### Interpretation

* **Positive covariance** → Both variables tend to increase together.
* **Negative covariance** → One variable tends to increase when the other decreases.
* **Covariance near zero** → Little linear co-movement.

For this business problem, if covariance is positive, it suggests that higher advertising spending tends to occur with higher daily sales.

However, covariance depends on the units and scale of the variables, so its magnitude is difficult to compare across different datasets.

---

# 2. Using Correlation

**Correlation** measures both the **direction and strength of the linear relationship** between two variables.

The Pearson correlation coefficient ranges from:

$$
-1 \leq r \leq 1
$$

### Interpretation

|      Correlation | Interpretation                   |
| ---------------: | -------------------------------- |
| \(r \approx +1\) | Strong positive relationship     |
|  \(r \approx 0\) | Little or no linear relationship |
| \(r \approx -1\) | Strong negative relationship     |

Correlation is easier to interpret than covariance because it is **unit-free**.

---

# Python Program

```python
import numpy as np

# Given data
advertising_spend = [200, 250, 300, 400, 500]
daily_sales = [2200, 2450, 2750, 3200, 4000]

# Calculate covariance
covariance = np.cov(advertising_spend, daily_sales)[0, 1]

# Calculate correlation
correlation = np.corrcoef(advertising_spend, daily_sales)[0, 1]

# Display results
print("Covariance:", covariance)
print("Correlation Coefficient:", correlation)
```

## Output

```text
Covariance: 22500.0
Correlation Coefficient: 0.994...
```

Therefore:

$$
\boxed{Covariance=22500}
$$

$$
\boxed{Correlation\approx0.994}
$$

---

# Interpretation of the Result

The correlation coefficient is approximately:

$$
r=0.994
$$

This value is very close to **+1**.

Therefore, there is a **very strong positive linear relationship** between advertising spend and daily sales in this dataset.

In simple terms:

> As advertising spend increases, daily sales also tend to increase.

For example:

```text
Advertising Spend     Daily Sales
      200   ─────────→   2200
      250   ─────────→   2450
      300   ─────────→   2750
      400   ─────────→   3200
      500   ─────────→   4000
```

The data shows a clear upward pattern.

---

# Business Interpretation

The marketing team can use this analysis as an initial indication that **higher advertising expenditure is associated with higher daily sales**.

For example:

* Spending ₹200 → Sales = ₹2,200
* Spending ₹300 → Sales = ₹2,750
* Spending ₹500 → Sales = ₹4,000

This suggests that advertising may be an important factor associated with sales performance.

However, **correlation does not prove causation**.

Other factors could also influence daily sales, such as:

* Discounts
* Seasonal demand
* Product availability
* Competitor pricing
* Holidays
* Website traffic
* Promotions
* Customer behavior

Therefore, the company should use additional analysis, such as **regression analysis or controlled experiments (A/B tests)**, before concluding that increasing advertising spend directly causes higher sales.
